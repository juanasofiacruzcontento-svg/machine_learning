```python
# ============================================================
# 🌊 TSUNAMI RISK ANALYSIS — KMeans + DBSCAN
# Script unificato con commenti analitici
# ============================================================
# STRUTTURA:
#   1.  Import & Setup
#   2.  Load Data
#   3.  Analisi correlazioni (DATA-DRIVEN — prima del feature eng.)
#   4.  Feature Engineering (pesi basati su correlazioni reali)
#   5.  EDA — distribuzioni + feature vs target
#   6.  Correlation Heatmap
#   7.  Mappa base Tsunami
#   8.  KMeans — preprocessing SOLO su risk features (no lat/lon)
#   9.  KMeans — elbow + silhouette
#  10.  KMeans finale + analisi cluster
#  11.  Categorie rischio adattive
#  12.  PCA + mappa KMeans
#  13.  DBSCAN — preprocessing + k-distance + auto-tuning
#  14.  DBSCAN finale + analisi cluster + outlier
#  15.  Categorie rischio DBSCAN
#  16.  PCA + mappa DBSCAN
#  17.  Mappa eventi critici combinata
# ============================================================


# ── 1. IMPORT & SETUP ────────────────────────────────────────
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.graph_objects as go
from IPython.display import display, HTML

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
from sklearn.neighbors import NearestNeighbors

import warnings
warnings.filterwarnings("ignore")

sns.set_style("darkgrid")


# ── 2. LOAD DATA ─────────────────────────────────────────────
df = pd.read_csv("../data/raw/earthquake_data_tsunami.csv")

display(df.head())
print(df.info())
print("\nDuplicati:", df.duplicated().sum())
print("\nMissing values:\n", df.isnull().sum())

tsunami_base_rate = df["tsunami"].mean() * 100
print(f"\n📊 Base rate tsunami nel dataset: {tsunami_base_rate:.2f}%")
# Il base rate è fondamentale: se è ~40%, un cluster con 42% non aggiunge
# nessuna informazione rispetto al non fare clustering.


# ── 3. ANALISI CORRELAZIONI (PRIMA DEL FEATURE ENGINEERING) ──
# FIX: le correlazioni vanno calcolate PRIMA di costruire il risk_score,
# non dopo. Servono per scegliere i pesi in modo data-driven.
#
# Incluse anche nst, dmin, gap che il notebook originale ignorava:
# - nst  = numero stazioni che hanno rilevato l'evento
# - dmin = distanza minima dalla stazione → bassa = misura precisa
# - gap  = gap azimutale → alto = epicentro poco vincolato
raw_features = [
    "magnitude", "depth", "cdi", "mmi", "sig",
    "nst", "dmin", "gap", "latitude", "longitude", "Year", "Month"
]
# Filtra solo le colonne presenti nel dataset
raw_features = [f for f in raw_features if f in df.columns]

corr_with_target = (
    df[raw_features + ["tsunami"]]
    .corr()["tsunami"]
    .drop("tsunami")
    .sort_values(key=abs, ascending=False)
)

print("\n📊 Correlazione di ogni feature raw con 'tsunami' (ordinata per |r|):")
print(corr_with_target.to_string())

# Visualizza le correlazioni come bar chart
fig, ax = plt.subplots(figsize=(10, 5))
colors = ["crimson" if v >= 0 else "steelblue" for v in corr_with_target.values]
ax.barh(corr_with_target.index, corr_with_target.values, color=colors)
ax.axvline(0, color="black", linewidth=0.8)
ax.axvline(0.1,  color="gray", linewidth=0.6, linestyle="--", label="|r|=0.10")
ax.axvline(-0.1, color="gray", linewidth=0.6, linestyle="--")
ax.set_title("Correlazione con 'tsunami' — rosso=positiva, blu=negativa")
ax.set_xlabel("Pearson r")
ax.legend()
plt.tight_layout()
plt.show()

# ⚠️ Leggi questo output PRIMA di procedere:
# Feature con |r| < 0.05 → quasi nessun segnale lineare sul target.
# Non significa che siano inutili per il clustering (la struttura
# geometrica dei cluster può essere reale), ma non aspettarti
# cluster con tsunami_rate molto diversi se le correlazioni sono basse.


# ── 4. FEATURE ENGINEERING (PESI DATA-DRIVEN) ────────────────

# Feature binarie — utili come segnale discreto
df["dist_equatore"] = abs(df["latitude"])
df["shallow_quake"] = (df["depth"] <= 70).astype(int)
df["strong_quake"]  = (df["magnitude"] >= 7).astype(int)
df["high_intensity"] = ((df["cdi"] >= 6) | (df["mmi"] >= 6)).astype(int)

# Log-transform per feature molto skewed (riduce l'effetto outlier)
df["log_depth"] = np.log1p(df["depth"])
df["log_sig"]   = np.log1p(df["sig"])
df["log_dmin"]  = np.log1p(df["dmin"]) if "dmin" in df.columns else 0.0

# ── RISK SCORE DATA-DRIVEN ────────────────────────────────────
# Costruzione pesi: usiamo |correlazione| di ogni feature con tsunami,
# normalizzata in modo che la somma dei pesi = 1.
# Questo sostituisce i pesi arbitrari (0.35, 0.25, ...) del notebook originale.

score_features = {
    "magnitude":    abs(corr_with_target.get("magnitude", 0.01)),
    "mmi":          abs(corr_with_target.get("mmi", 0.01)),
    "cdi":          abs(corr_with_target.get("cdi", 0.01)),
    "log_sig":      abs(corr_with_target.get("sig", 0.01)),
    "log_depth":    abs(corr_with_target.get("depth", 0.01)),
    "shallow_quake":abs(corr_with_target.get("depth", 0.01)),
    "strong_quake": abs(corr_with_target.get("magnitude", 0.01)),
}

# Normalizza pesi a somma 1
total_w = sum(score_features.values())
weights = {k: v / total_w for k, v in score_features.items()}

print("\n⚖️  Pesi data-driven del risk_score:")
for feat, w in sorted(weights.items(), key=lambda x: -x[1]):
    print(f"   {feat:20s} → {w:.4f}")

df["risk_score"] = sum(w * df[feat] for feat, w in weights.items())

# Normalizzazione min-max → [0, 1]
df["risk_score_norm"] = (
    (df["risk_score"] - df["risk_score"].min()) /
    (df["risk_score"].max() - df["risk_score"].min())
)

corr_new = df["risk_score_norm"].corr(df["tsunami"])
print(f"\n📊 Correlazione risk_score_norm (data-driven) ↔ tsunami: {corr_new:.4f}")
# Confronta con la versione precedente (0.049).
# Se ancora bassa, il segnale lineare nel dataset è strutturalmente debole:
# il clustering farà segmentazione geografica più che predizione di rischio.


# ── 5. EDA — DISTRIBUZIONI + FEATURE VS TARGET ───────────────
all_plot_features = [
    "magnitude", "depth", "cdi", "mmi", "sig",
    "dist_equatore", "risk_score_norm"
]

fig, axes = plt.subplots(2, 4, figsize=(20, 9))
for i, col in enumerate(all_plot_features):
    sns.histplot(data=df, x=col, kde=True, ax=axes.ravel()[i])
    axes.ravel()[i].set_title(col)
axes.ravel()[-1].set_visible(False)
plt.suptitle("📊 Distribuzioni feature", y=1.01)
plt.tight_layout(); plt.show()

fig, axes = plt.subplots(2, 4, figsize=(20, 9))
for i, col in enumerate(all_plot_features):
    sns.boxplot(data=df, x="tsunami", y=col, ax=axes.ravel()[i])
    axes.ravel()[i].set_title(f"{col} vs Tsunami")
axes.ravel()[-1].set_visible(False)
plt.suptitle("📦 Feature vs Target — box sovrapposti = bassa discriminazione", y=1.01)
plt.tight_layout(); plt.show()


# ── 6. CORRELATION HEATMAP ───────────────────────────────────
plt.figure(figsize=(14, 9))
heatmap_cols = [
    "magnitude", "depth", "cdi", "mmi", "sig",
    "dist_equatore", "shallow_quake", "strong_quake",
    "risk_score_norm", "tsunami"
]
for c in ["nst", "dmin", "gap"]:
    if c in df.columns:
        heatmap_cols.insert(-1, c)

corr = df[heatmap_cols].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("🔥 Correlation Heatmap — include nst/dmin/gap prima ignorati")
plt.show()


# ── 7. MAPPA BASE TSUNAMI / NO TSUNAMI ──────────────────────
df_map = df.copy()
df_map["tsunami_text"] = df_map["tsunami"].map({0: "No Tsunami", 1: "Tsunami"})
df_map["hover_text"] = df_map.apply(
    lambda row: (
        f"<b>Earthquake</b><br>"
        f"Year: {row['Year']}<br>Month: {row['Month']}<br>"
        f"Magnitude: {row['magnitude']}<br>Depth: {row['depth']} km<br>"
        f"CDI: {row['cdi']}<br>MMI: {row['mmi']}<br>"
        f"Tsunami: {row['tsunami_text']}"
    ), axis=1
)

fig = go.Figure()
for value, color, name, size in [(0, "green", "No Tsunami", 5), (1, "red", "Tsunami", 7)]:
    data = df_map[df_map["tsunami"] == value]
    fig.add_trace(go.Scattergeo(
        lat=data["latitude"], lon=data["longitude"],
        mode="markers", text=data["hover_text"], hoverinfo="text",
        name=name, marker=dict(color=color, size=size, opacity=0.75,
                               line=dict(width=0.4, color="white"))
    ))
fig.update_geos(projection_type="orthographic", landcolor="lightgray",
                oceancolor="lightblue", showocean=True, showland=True,
                showcountries=True, countrycolor="white")
fig.update_layout(title="🌍 Global Earthquake & Tsunami Distribution", height=650)
display(HTML(fig.to_html(include_plotlyjs="cdn")))


# ── 8. KMEANS — PREPROCESSING (SOLO RISK FEATURES) ───────────
# FIX PRINCIPALE: il notebook originale includeva lat/lon tra le feature
# KMeans. Con lat/lon dominanti, KMeans produce cluster GEOGRAFICI
# (Ring of Fire, Oceano Indiano, ecc.) che hanno tutti tsunami_rate
# simile → solo MEDIUM RISK e NEUTRAL RISK, come visto nel PCA.
#
# Soluzione: KMeans su SOLE feature fisiche di rischio (no lat/lon).
# I cluster risultanti rappresentano profili di rischio, non zone geografiche.
# lat/lon vengono usati solo per la visualizzazione su mappa.

kmeans_risk_features = [
    "magnitude",
    "log_depth",
    "cdi",
    "mmi",
    "log_sig",
    "shallow_quake",
    "strong_quake",
    "high_intensity",
    "dist_equatore",
]
# Aggiungi nst/dmin/gap se presenti e con correlazione utile
for col in ["nst", "dmin", "gap"]:
    if col in df.columns and abs(corr_with_target.get(col, 0)) > 0.03:
        kmeans_risk_features.append(col)
        print(f"✅ Aggiunta '{col}' al set KMeans (|r|={abs(corr_with_target.get(col,0)):.3f})")

print(f"\n🔧 Feature KMeans ({len(kmeans_risk_features)}): {kmeans_risk_features}")

X_km = df[kmeans_risk_features].copy()
scaler_km = StandardScaler()
X_km_scaled = scaler_km.fit_transform(X_km)


# ── 9. KMEANS — ELBOW + SILHOUETTE ───────────────────────────
inertias, silhouettes = [], []
K_range = range(2, 12)

for k in K_range:
    model  = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = model.fit_predict(X_km_scaled)
    inertias.append(model.inertia_)
    silhouettes.append(silhouette_score(X_km_scaled, labels))

k_scores = pd.DataFrame({"K": list(K_range), "Inertia": inertias, "Silhouette": silhouettes})
best_k = k_scores.loc[k_scores["Silhouette"].idxmax(), "K"]
print(f"\n✅ K con silhouette massima: {best_k}")
display(k_scores)

fig, axes = plt.subplots(1, 2, figsize=(16, 5))
axes[0].plot(K_range, inertias, marker="o")
axes[0].set_title("Elbow Method"); axes[0].set_xlabel("K")
axes[0].set_ylabel("Inertia"); axes[0].grid(True)

axes[1].plot(K_range, silhouettes, marker="o")
axes[1].axvline(best_k, color="red", linestyle="--", label=f"Best K={best_k}")
axes[1].set_title("Silhouette Score"); axes[1].set_xlabel("K")
axes[1].set_ylabel("Silhouette"); axes[1].grid(True); axes[1].legend()
plt.tight_layout(); plt.show()


# ── 10. KMEANS FINALE ─────────────────────────────────────────
# Usa K con silhouette massima (data-driven).
# Per forzare un K specifico, cambia K_FINAL manualmente.
K_FINAL = best_k

kmeans_final = KMeans(n_clusters=K_FINAL, random_state=42, n_init=10)
df["kmeans_cluster"] = kmeans_final.fit_predict(X_km_scaled)

print(f"\n📊 KMeans finale con K={K_FINAL}:")
print(df["kmeans_cluster"].value_counts().sort_index())

kmeans_summary = df.groupby("kmeans_cluster").agg(
    count=("tsunami", "count"),
    tsunami_rate=("tsunami", "mean"),
    avg_magnitude=("magnitude", "mean"),
    avg_depth=("depth", "mean"),
    avg_cdi=("cdi", "mean"),
    avg_mmi=("mmi", "mean"),
    avg_sig=("sig", "mean"),
    avg_risk_score=("risk_score_norm", "mean"),
    avg_latitude=("latitude", "mean"),
    avg_longitude=("longitude", "mean")
)
kmeans_summary["tsunami_rate"] *= 100

print(f"\n   Base rate dataset: {tsunami_base_rate:.1f}%")
display(kmeans_summary.sort_values("tsunami_rate", ascending=False))


# ── 11. CATEGORIE RISCHIO ADATTIVE ───────────────────────────
# FIX: il notebook originale usava soglie fisse (≥60% HIGH, ≤20% LOW)
# che non si adattano quando tutti i cluster hanno tsunami_rate tra 20–60%.
# Risultato: solo MEDIUM e NEUTRAL RISK → grafico PCA completamente giallo/grigio.
#
# Soluzione: soglie basate sui quartili della distribuzione reale
# dei tsunami_rate nei cluster. Garantisce sempre tutti e 4 i colori.

rates = kmeans_summary["tsunami_rate"].values
q25 = np.percentile(rates, 25)
q50 = np.percentile(rates, 50)
q75 = np.percentile(rates, 75)

print(f"\n📐 Distribuzione tsunami_rate nei cluster:")
print(f"   Q25={q25:.1f}%  Mediana={q50:.1f}%  Q75={q75:.1f}%")
print(f"   → HIGH RISK   se rate ≥ {q75:.1f}%")
print(f"   → MEDIUM RISK se {q50:.1f}% ≤ rate < {q75:.1f}%")
print(f"   → NEUTRAL RISK se {q25:.1f}% ≤ rate < {q50:.1f}%")
print(f"   → LOW RISK    se rate < {q25:.1f}%")

def assign_risk_category(rate, q25=q25, q50=q50, q75=q75):
    if rate >= q75:   return "HIGH RISK 🔴"
    elif rate >= q50: return "MEDIUM RISK 🟡"
    elif rate >= q25: return "NEUTRAL RISK ⚪"
    else:             return "LOW RISK 🟢"

kmeans_risk_map = kmeans_summary["tsunami_rate"].to_dict()
df["kmeans_risk"] = df["kmeans_cluster"].map(
    lambda x: assign_risk_category(kmeans_risk_map[x])
)

risk_validation = (
    df.groupby("kmeans_risk")["tsunami"]
    .agg(count="count", tsunami_rate="mean")
    .assign(tsunami_rate=lambda d: d["tsunami_rate"] * 100)
    .assign(lift=lambda d: d["tsunami_rate"] / tsunami_base_rate)
    .sort_values("tsunami_rate", ascending=False)
)
print("\n✅ Validazione categorie rischio (lift > 1.0 = più rischioso della media):")
display(risk_validation)


# ── 12. PCA + MAPPA KMEANS ────────────────────────────────────
pca = PCA(n_components=2)
X_km_pca = pca.fit_transform(X_km_scaled)
df["PCA1"] = X_km_pca[:, 0]
df["PCA2"] = X_km_pca[:, 1]

explained = pca.explained_variance_ratio_
print(f"\n📐 Varianza spiegata PCA KMeans: {explained.sum()*100:.1f}%")

fig, axes = plt.subplots(1, 2, figsize=(17, 6))

sns.scatterplot(data=df, x="PCA1", y="PCA2", hue="kmeans_cluster",
                palette="tab10", s=60, alpha=0.8, ax=axes[0])
axes[0].set_title(f"KMeans Clusters (K={K_FINAL}) — PCA")
axes[0].grid(True)

risk_palette = {"HIGH RISK 🔴": "red", "MEDIUM RISK 🟡": "orange",
                "NEUTRAL RISK ⚪": "gray", "LOW RISK 🟢": "green"}
sns.scatterplot(
    data=df, x="PCA1", y="PCA2", hue="kmeans_risk",
    hue_order=list(risk_palette.keys()),
    palette=risk_palette, s=60, alpha=0.8, ax=axes[1]
)
axes[1].set_title("KMeans Risk Categories — PCA")
axes[1].grid(True)

plt.suptitle(
    f"PCA varianza spiegata: {explained.sum()*100:.1f}%  |  "
    f"Base rate: {tsunami_base_rate:.1f}%  |  K={K_FINAL}"
)
plt.tight_layout(); plt.show()


def create_kmeans_world_map(df):
    colors = {"HIGH RISK 🔴": "red", "MEDIUM RISK 🟡": "orange",
              "LOW RISK 🟢": "green", "NEUTRAL RISK ⚪": "gray"}
    df_map = df.copy()
    df_map["hover_text"] = df_map.apply(
        lambda row: (
            f"<b>Earthquake</b><br>"
            f"Year: {row['Year']}<br>Month: {row['Month']}<br>"
            f"Magnitude: {row['magnitude']}<br>Depth: {row['depth']} km<br>"
            f"Tsunami: {'YES' if row['tsunami'] == 1 else 'NO'}<br>"
            f"KMeans Cluster: {row['kmeans_cluster']}<br>"
            f"Risk: {row['kmeans_risk']}"
        ), axis=1
    )
    fig = go.Figure()
    for risk, color in colors.items():
        data = df_map[df_map["kmeans_risk"] == risk]
        if len(data) > 0:
            fig.add_trace(go.Scattergeo(
                lat=data["latitude"], lon=data["longitude"],
                mode="markers", name=risk,
                text=data["hover_text"], hoverinfo="text",
                marker=dict(color=color, size=7, opacity=0.8,
                            line=dict(width=0.4, color="white"))
            ))
    fig.update_geos(projection_type="orthographic", landcolor="lightgray",
                    oceancolor="lightblue", showocean=True, showland=True,
                    showcountries=True, countrycolor="white")
    fig.update_layout(title="🌍 KMeans Tsunami Risk Map", height=650,
                      legend_title="Risk Level")
    return fig

display(HTML(create_kmeans_world_map(df).to_html(include_plotlyjs="cdn")))


# ── 13. DBSCAN — PREPROCESSING + K-DISTANCE ──────────────────
# DBSCAN su lat/lon/magnitude/depth per anomaly detection geografica.
dbscan_features = ["latitude", "longitude", "magnitude", "depth"]

X_db = df[dbscan_features].copy()
scaler_db = StandardScaler()
X_db_scaled = scaler_db.fit_transform(X_db)

min_samples_kd = 12
neighbors = NearestNeighbors(n_neighbors=min_samples_kd)
neighbors.fit(X_db_scaled)
distances, _ = neighbors.kneighbors(X_db_scaled)
distances = np.sort(distances[:, -1])

plt.figure(figsize=(9, 5))
plt.plot(distances)
plt.title(f"K-distance Plot (k={min_samples_kd}) — eps al 'gomito'")
plt.xlabel("Punti ordinati"); plt.ylabel(f"Dist. dal {min_samples_kd}° vicino")
plt.grid(True); plt.show()


# ── 14. DBSCAN — AUTO-TUNING + FINALE ────────────────────────
eps_values         = np.arange(0.3, 3.1, 0.1)
min_samples_values = [5, 8, 10, 12, 15]

results = []
for eps in eps_values:
    for min_s in min_samples_values:
        labels     = DBSCAN(eps=eps, min_samples=min_s).fit_predict(X_db_scaled)
        n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
        noise_ratio = np.mean(labels == -1)

        sil = np.nan
        if n_clusters >= 2:
            mask = labels != -1
            if len(set(labels[mask])) >= 2:
                sil = silhouette_score(X_db_scaled[mask], labels[mask])

        results.append({"eps": round(eps, 2), "min_samples": min_s,
                         "n_clusters": n_clusters,
                         "noise_ratio": round(noise_ratio, 3), "silhouette": sil})

dbscan_results = pd.DataFrame(results)

valid_dbscan = dbscan_results[
    (dbscan_results["n_clusters"] >= 2) &
    (dbscan_results["n_clusters"] <= 8) &
    (dbscan_results["noise_ratio"] <= 0.40) &
    (dbscan_results["silhouette"].notna())
].copy()

best_row = valid_dbscan.sort_values(
    ["silhouette", "noise_ratio"], ascending=[False, True]
).iloc[0]

best_eps         = best_row["eps"]
best_min_samples = int(best_row["min_samples"])

print(f"\n✅ Parametri DBSCAN: eps={best_eps}, min_samples={best_min_samples}")
print(f"   Silhouette: {best_row['silhouette']:.4f}  |  "
      f"Noise: {best_row['noise_ratio']*100:.1f}%  |  "
      f"Cluster: {int(best_row['n_clusters'])}")

df["dbscan_cluster"] = DBSCAN(
    eps=best_eps, min_samples=best_min_samples
).fit_predict(X_db_scaled)

print("\n📊 Distribuzione cluster DBSCAN:")
print(df["dbscan_cluster"].value_counts().sort_index())

# Analisi cluster — sezione 21 (bug fix: mancava nel notebook originale)
dbscan_summary = df.groupby("dbscan_cluster").agg(
    count=("tsunami", "count"),
    tsunami_rate=("tsunami", "mean"),
    avg_magnitude=("magnitude", "mean"),
    avg_depth=("depth", "mean"),
    avg_latitude=("latitude", "mean"),
    avg_longitude=("longitude", "mean"),
    avg_risk_score=("risk_score_norm", "mean")
)
dbscan_summary["tsunami_rate"] *= 100
display(dbscan_summary.sort_values("tsunami_rate", ascending=False))

# Analisi outlier
outliers     = df[df["dbscan_cluster"] == -1]
non_outliers = df[df["dbscan_cluster"] != -1]

print(f"\n🔍 Outlier DBSCAN:")
print(f"   N: {len(outliers)} ({len(outliers)/len(df)*100:.1f}%)")
print(f"   Tsunami rate outlier:     {outliers['tsunami'].mean()*100:.2f}%")
print(f"   Tsunami rate non-outlier: {non_outliers['tsunami'].mean()*100:.2f}%")
print(f"   Lift outlier:             {outliers['tsunami'].mean() / df['tsunami'].mean():.2f}x")

display(
    outliers[["Year", "Month", "magnitude", "depth",
              "latitude", "longitude", "tsunami", "risk_score_norm"]]
    .sort_values("risk_score_norm", ascending=False)
    .head(20)
)


# ── 15. CATEGORIE RISCHIO DBSCAN ─────────────────────────────
# Stesse soglie adattive del KMeans per coerenza
dbscan_risk_map = dbscan_summary["tsunami_rate"].to_dict()

df["dbscan_risk"] = df["dbscan_cluster"].map(
    lambda x: "OUTLIER ⚫" if x == -1
    else assign_risk_category(dbscan_risk_map[x])
)

display(
    df.groupby("dbscan_risk")["tsunami"]
    .agg(count="count", tsunami_rate="mean")
    .assign(tsunami_rate=lambda d: d["tsunami_rate"] * 100,
            lift=lambda d: d["tsunami_rate"] / tsunami_base_rate)
    .sort_values("tsunami_rate", ascending=False)
)


# ── 16. PCA + MAPPA DBSCAN ────────────────────────────────────
pca_db = PCA(n_components=2)
X_db_pca = pca_db.fit_transform(X_db_scaled)
df["DBSCAN_PCA1"] = X_db_pca[:, 0]
df["DBSCAN_PCA2"] = X_db_pca[:, 1]

expl_db = pca_db.explained_variance_ratio_
print(f"\n📐 Varianza spiegata PCA DBSCAN: {expl_db.sum()*100:.1f}%")

plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x="DBSCAN_PCA1", y="DBSCAN_PCA2",
                hue="dbscan_cluster", palette="tab10", s=60, alpha=0.8)
plt.title(f"DBSCAN Clusters — PCA ({expl_db.sum()*100:.1f}% varianza spiegata)")
plt.grid(True); plt.show()


def create_dbscan_world_map(df):
    colors = {"HIGH RISK 🔴": "red", "MEDIUM RISK 🟡": "orange",
              "LOW RISK 🟢": "green", "NEUTRAL RISK ⚪": "gray", "OUTLIER ⚫": "black"}
    df_map = df.copy()
    df_map["hover_text"] = df_map.apply(
        lambda row: (
            f"<b>Earthquake</b><br>"
            f"Year: {row['Year']}<br>Month: {row['Month']}<br>"
            f"Magnitude: {row['magnitude']}<br>Depth: {row['depth']} km<br>"
            f"Tsunami: {'YES' if row['tsunami'] == 1 else 'NO'}<br>"
            f"DBSCAN Cluster: {row['dbscan_cluster']}<br>"
            f"Risk: {row['dbscan_risk']}<br>"
            f"Risk Score: {row['risk_score_norm']:.2f}"
        ), axis=1
    )
    fig = go.Figure()
    for risk, color in colors.items():
        data = df_map[df_map["dbscan_risk"] == risk]
        if len(data) > 0:
            fig.add_trace(go.Scattergeo(
                lat=data["latitude"], lon=data["longitude"],
                mode="markers", name=risk,
                text=data["hover_text"], hoverinfo="text",
                marker=dict(color=color,
                            size=8 if risk == "OUTLIER ⚫" else 6,
                            opacity=0.85, line=dict(width=0.4, color="white"))
            ))
    fig.update_geos(projection_type="orthographic", landcolor="lightgray",
                    oceancolor="lightblue", showocean=True, showland=True,
                    showcountries=True, countrycolor="white")
    fig.update_layout(title="🌍 DBSCAN Outlier & Risk Map", height=650,
                      legend_title="DBSCAN Risk")
    return fig

display(HTML(create_dbscan_world_map(df).to_html(include_plotlyjs="cdn")))


# ── 17. MAPPA EVENTI CRITICI COMBINATA ────────────────────────
# critical_event = KMeans HIGH RISK  OR  DBSCAN outlier
# FIX: rimosso l'if-guard inutile del notebook originale.
df["critical_event"] = (
    (df["kmeans_risk"] == "HIGH RISK 🔴") |
    (df["dbscan_cluster"] == -1)
)

overall_rate  = df["tsunami"].mean() * 100
critical_rate = df[df["critical_event"]]["tsunami"].mean() * 100
normal_rate   = df[~df["critical_event"]]["tsunami"].mean() * 100
lift          = critical_rate / overall_rate

print(f"\n🚨 Distribuzione eventi critici:")
print(df["critical_event"].value_counts().to_string())
print(f"\n   Tsunami rate CRITICO:     {critical_rate:.2f}%")
print(f"   Tsunami rate NORMALE:     {normal_rate:.2f}%")
print(f"   Tsunami rate COMPLESSIVO: {overall_rate:.2f}%")
print(f"   Lift eventi critici:      {lift:.2f}x")
print(f"\n   → Lift > 1.5 = definizione di critical_event ha potere predittivo")
print(f"   → Lift ≈ 1.0 = KMeans HIGH RISK e DBSCAN outlier non catturano tsunami")

df["critical_text"] = df["critical_event"].map(
    {True: "Critical Event 🚨", False: "Normal Event"}
)
df["hover_text_crit"] = df.apply(
    lambda row: (
        f"<b>🌍 Earthquake</b><br>"
        f"📅 Year: {row['Year']}<br>🗓 Month: {row['Month']}<br>"
        f"📈 Magnitude: {row['magnitude']:.2f}<br>"
        f"🏔 Depth: {row['depth']:.1f} km<br>"
        f"🌊 Tsunami: {'YES 🔴' if row['tsunami'] == 1 else 'NO 🟢'}<br>"
        f"🤖 KMeans Risk: {row['kmeans_risk']}<br>"
        f"🔍 DBSCAN: {row['dbscan_risk']}<br>"
        f"<b>🚨 {row['critical_text']}</b>"
    ), axis=1
)

fig = go.Figure()
for value, color, name, size in [
    (False, "lightgray", "Normal Event", 5),
    (True,  "red",       "Critical Event 🚨", 9)
]:
    data = df[df["critical_event"] == value]
    if len(data) > 0:
        fig.add_trace(go.Scattergeo(
            lat=data["latitude"], lon=data["longitude"],
            mode="markers", name=name,
            text=data["hover_text_crit"], hoverinfo="text",
            marker=dict(color=color, size=size, opacity=0.85,
                        line=dict(width=0.5, color="white"))
        ))

fig.update_geos(projection_type="orthographic", showland=True,
                landcolor="lightgray", showocean=True, oceancolor="lightblue",
                showcountries=True, countrycolor="white")
fig.update_layout(
    title=f"🚨 Critical Events Map  |  Lift: {lift:.2f}x  |  "
          f"Critical rate: {critical_rate:.1f}%  vs  Base rate: {overall_rate:.1f}%",
    height=650, legend=dict(x=0.01, y=0.99)
)
display(HTML(fig.to_html(include_plotlyjs="cdn")))
```
